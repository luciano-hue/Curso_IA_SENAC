## 7. Tratamento de texto e padronização

### Exercício 13
Padronize a coluna `bolsista` para ficar apenas com:
- `Sim`
- `Não`

Dica: existem valores como `SIM`, `sim`, `Nao`.

### Exercício 14
Padronize a coluna `uf` para letras maiúsculas.

### Exercício 15
Padronize a coluna `cidade` no formato título.
Exemplo:
- `brasília` → `Brasília`


```python
df = pd.read_excel("base_exercicios_alunos_pandas.xlsx")

# 1. Preenche valores NaN na coluna 'bolsista' com 'Não'
# Assumimos que NaN neste contexto significa que o aluno não é bolsista.
df["bolsista"] = df["bolsista"].fillna('Não')

# 2. Converte todas as entradas para minúsculas para padronização de case
df["bolsista"] = df["bolsista"].str.lower()

# 3. Mapeia as variações para 'Sim' ou 'Não' padronizados
df["bolsista"] = df["bolsista"].replace({
    "sim": "Sim",
    "nao": "Não",
    "não": "Não" # Garante que 'não' após lower() seja mapeado
})

display(df)
```

```python
#Exercício 14
#Padronize a coluna uf para letras maiúsculas.
df = pd.read_excel("base_exercicios_alunos_pandas.xlsx")
df["uf"] = df["uf"].str.upper()
display(df)
```

```python
#Padronize a coluna cidade no formato título. Exemplo:

#brasília → Brasília
df = pd.read_excel("base_exercicios_alunos_pandas.xlsx")
df["cidade"] = df["cidade"].str.title()
display(df)
```

## 8. Criando novas colunas

### Exercício 16
Crie uma coluna chamada `media` com a média entre `nota1` e `nota2`.

### Exercício 17
Crie uma coluna chamada `aprovado`:
- `True` para média maior ou igual a **7**
- `False` caso contrário

### Exercício 18
Crie uma coluna chamada `situacao_faltas`:
- `Alto risco` para mais de 10 faltas
- `Atenção` para 6 a 10 faltas
- `Ok` para até 5 faltas


```python
# Resolva aqui
#Exercício 16
#Crie uma coluna chamada media com a média entre nota1 e nota2.
df = pd.read_excel("base_exercicios_alunos_pandas.xlsx")
df["media"] = (df["nota1"] + df["nota2"]) / 2
display(df)

```

```python
#Exercício 17
#Crie uma coluna chamada aprovado:

#True para média maior ou igual a 7
#False caso contrário
df = pd.read_excel("base_exercicios_alunos_pandas.xlsx")
df["media"] = (df["nota1"] + df["nota2"]) / 2 # Garante que a coluna 'media' seja criada
df["aprovado"] = df["media"] >= 7
display(df)
```

```python
#Exercício 18
#Crie uma coluna chamada situacao_faltas:

#Alto risco para mais de 10 faltas
#Atenção para 6 a 10 faltas
#Ok para até 5 faltas
df = pd.read_excel("base_exercicios_alunos_pandas.xlsx")
df["situacao_faltas"] = df["faltas"].apply(lambda x: "Alto risco" if x > 10 else ("Atenção" if x >= 6 else "Ok"))
display(df)
```

## 13. Desafio final

### Exercício 31 — Desafio
Monte uma análise final respondendo:

1. Quantos alunos há em cada curso?
2. Qual curso tem a maior média de `nota1`?
3. Quantos alunos têm mais de 10 faltas?
4. Quantos alunos são bolsistas?
5. Há dados faltando? Em quais colunas?
6. Há duplicatas?
7. Gere um DataFrame final apenas com:
   - `nome`
   - `curso`
   - `media`
   - `aprovado`

Depois, ordene esse DataFrame pela `media` em ordem decrescente.


```python
# Resolva aqui

# Carrega o DataFrame novamente para garantir o estado inicial
df = pd.read_excel("base_exercicios_alunos_pandas.xlsx")

# Garante que as colunas 'bolsista' e 'uf' estejam padronizadas como nos exercícios anteriores
df["bolsista"] = df["bolsista"].fillna('Não')
df["bolsista"] = df["bolsista"].str.lower()
df["bolsista"] = df["bolsista"].replace({"sim": "Sim", "nao": "Não", "não": "Não"})
df["uf"] = df["uf"].str.upper()
df["cidade"] = df["cidade"].str.title()

# 1. Quantos alunos há em cada curso?
print("1. Quantos alunos há em cada curso?")
display(df["curso"].value_counts())
print("\n" + "-"*30 + "\n")

# Recalcula 'media' e 'aprovado' para garantir que estejam presentes e atualizados
df["media"] = (df["nota1"] + df["nota2"]) / 2
df["aprovado"] = df["media"] >= 7

# 2. Qual curso tem a maior média de nota1?
print("2. Qual curso tem a maior média de nota1?")
media_nota1_por_curso = df.groupby("curso")["nota1"].mean()
curso_maior_media_nota1 = media_nota1_por_curso.idxmax()
display(f"O curso com a maior média de nota1 é: {curso_maior_media_nota1} ({media_nota1_por_curso.max():.2f})")
print("\n" + "-"*30 + "\n")

# 3. Quantos alunos têm mais de 10 faltas?
print("3. Quantos alunos têm mais de 10 faltas?")
alunos_alto_risco_faltas = df[df["faltas"] > 10].shape[0]
display(f"Número de alunos com mais de 10 faltas: {alunos_alto_risco_faltas}")
print("\n" + "-"*30 + "\n")

# 4. Quantos alunos são bolsistas?
print("4. Quantos alunos são bolsistas?")
alunos_bolsistas = df[df["bolsista"] == "Sim"].shape[0]
display(f"Número de alunos bolsistas: {alunos_bolsistas}")
print("\n" + "-"*30 + "\n")

# 5. Há dados faltando? Em quais colunas?
print("5. Há dados faltando? Em quais colunas?")
dados_faltando = df.isnull().sum()
display(dados_faltando[dados_faltando > 0])
print("\n" + "-"*30 + "\n")

# 6. Há duplicatas?
print("6. Há duplicatas?")
duplicatas_count = df.duplicated().sum()
display(f"Há duplicatas na base de dados: {duplicatas_count > 0} (Total: {duplicatas_count})")
print("\n" + "-"*30 + "\n")

# 7. Gere um DataFrame final
print("7. Gerando DataFrame final (nome, curso, media, aprovado) e ordenando pela média:")
df_final = df[["nome", "curso", "media", "aprovado"]]
df_final = df_final.sort_values(by="media", ascending=False)
display(df_final.head(10)) # Mostra as 10 primeiras linhas do DataFrame final
```
